```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import math
pd.set_option('display.max_columns', None)
```

```python
trainFolder = os.fsencode("seasons/train")
testFolder = os.fsencode("seasons/test")

dataframes = []
for file in sorted(os.listdir(trainFolder)):
    filename = os.fsdecode(file)
    if filename.endswith(".csv"):
        df = pd.read_csv("seasons/train/" + filename)
        dataframes.append(df)

for file in sorted(os.listdir(testFolder)):
    filename = os.fsdecode(file)
    if filename.endswith(".csv"):
        df = pd.read_csv("seasons/test/" + filename)
        dataframes.append(df)

data = pd.concat(dataframes, axis=0)
```

```python
data = data[[
    "Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG", "FTR", "HS", "AS", "HST", "AST", "B365H", "B365D", "B365A"
]]
```

```python
scorelines = data[["FTHG", "FTAG"]]
```

```python
goalsScored = pd.concat([scorelines.FTAG, scorelines.FTHG])
```

```python
np.mean(goalsScored)
```

```python
np.var(goalsScored)
```

```python
poissonProbs = [(math.exp(-np.mean(goalsScored)) * np.mean(goalsScored)**i) / math.factorial(i) for i in range(int(max(goalsScored))+1)]
```

```python
# Get counts per unique value
values, counts = np.unique(goalsScored, return_counts=True)

# Normalize if you want probabilities
counts = counts / counts.sum()

# Make bar plot
plt.bar(values, counts, width=0.9, align='center', edgecolor='black', label="Observed Frequencies")
plt.scatter([i for i in range(int(max(goalsScored))+1)], poissonProbs, color="red", label="Expected Poisson Frequencies")

# Label the center values
plt.xticks([i for i in range(int(max(values)) + 1)])
plt.xlabel("Goals Scored")
plt.ylabel("Probability")
plt.title("Distribution of Goals Scored")
plt.legend()
plt.title("Number of Goals Scored by Teams in English Premier League Matches between 2006-2025", wrap=True)
plt.savefig('goalsScored.pdf', bbox_inches='tight',)
```

```python
scorelines = scorelines.dropna()
scorelines["Score"] = (
    scorelines["FTHG"].astype(int).astype(str)
    + "-" +
    scorelines["FTAG"].astype(int).astype(str)
)
```

```python
scorelines.head()
```

```python
scorelines.groupby(by="Score").count()
```

```python
freq_table = pd.crosstab(scorelines["FTHG"].astype(int),
                         scorelines["FTAG"].astype(int))
```

```python
freq_table
```

```python
# Export to LaTeX
latex = freq_table.to_latex(index=True,
                       caption="Empirical joint frequency table of Home and Away goals",
                       label="tab:jointfreq",
                       column_format="r" + "c" * freq_table.shape[1],
                       bold_rows=True)  # bolds the index (home goals)
print(latex)
```
