```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
headsProb = 0.55
alpha = 0.25

flatHist = [100]
kellyHist = [100]
fractionalKellyHist = [100]
fig, ax = plt.subplots()

np.random.seed(1038)

for i in range(5):
    flatBankroll = 100
    kellyBankroll = 100
    fractionalBankroll = 100
    
    flatHist = [100]
    kellyHist = [100]
    fractionalKellyHist = [100]
    for j in range(100):
        coinIsHeads = 1 if np.random.rand() < headsProb else 0
        
        flatStake = 10
        kellyStake = 0.1*kellyBankroll
        fractionalStake = 0.1*fractionalBankroll*alpha

        if flatBankroll - flatStake > 0:
            flatBankroll += coinIsHeads*2*flatStake - flatStake
        else:
            flatBankroll = 0
        if kellyBankroll - kellyStake > 0:
            kellyBankroll += coinIsHeads*2*kellyStake - kellyStake
        else:
            kellyBankroll = 0
        if fractionalBankroll - fractionalStake > 0:
            fractionalBankroll += coinIsHeads*2*fractionalStake - fractionalStake
        else:
            fractionalBankroll = 0
        flatHist.append(flatBankroll)
        kellyHist.append(kellyBankroll)
        fractionalKellyHist.append(fractionalBankroll)
    ax.plot(flatHist, color="blue", alpha=0.4)
    ax.plot(kellyHist, color="red", alpha=0.4)
    ax.plot(fractionalKellyHist, color="green", alpha=0.6)
ax.plot([], color="blue", alpha=0.4, label="Flat £10 Stakes")
ax.plot([], color="red", alpha=0.4, label="Kelly Stakes")
ax.plot([], color="green", alpha=0.4, label="Fractional Kelly Stakes")
ax.set_title("Performance of Staking Strategies over 100 bets")
ax.hlines(100, 0, 100, color="black", linestyle="dashed")
ax.legend()

plt.savefig("staking.jpg")
```
