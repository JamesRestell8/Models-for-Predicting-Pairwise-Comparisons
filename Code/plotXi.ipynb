```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math
import random
```

```python
x = np.arange(1, 1200)
y = np.exp(-0.0065*x)

fig, ax = plt.subplots()
ax.plot(x, y)
plt.xscale("log")
plt.title("$\\phi(t - t_k)$")
plt.xlabel("t - t_k")
plt.ylabel("$\\exp(-0.0065(t - t_k))$")
plt.savefig("phi.jpg")
```
