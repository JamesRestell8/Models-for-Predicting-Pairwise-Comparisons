```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import csv
import warnings
import os

warnings.simplefilter(action="ignore", category=FutureWarning)
warnings.simplefilter(action="ignore", category=UserWarning)
warnings.simplefilter("ignore", pd.errors.SettingWithCopyWarning)
```

```python
folder = os.fsencode("data/football")
dataframes = []
for file in sorted(os.listdir(folder)):
    filename = os.fsdecode(file)
    if filename.endswith(".csv"):
        df = pd.read_csv("data/football/" + filename)
        dataframes.append(df)

data = pd.concat(dataframes, axis=0)
```

```python
data = data[[
    "Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG", "FTR",
    "B365H", "B365D", "B365A", "AvgH", "AvgD", "AvgA", 
    "B365>2.5", "B365<2.5", "Avg>2.5", "Avg<2.5"
]]
```

```python
data.describe()
```

```python
data["1x2margin"] = (1/data["B365H"]) + (1/data["B365D"]) + (1/data["B365A"])
```

```python
data["OUmargin"] = (1/data["B365>2.5"]) + (1/data["B365<2.5"])
```
