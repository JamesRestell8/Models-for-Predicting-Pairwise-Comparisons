```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta
```

```python
# Define the distribution
a, b = 2, 1
x = np.linspace(0, 1, 1000)
pdf = beta.pdf(x, a, b)

# Plot
plt.figure(figsize=(6, 4))
plt.plot(x, pdf, label=r'$\mathrm{Beta}(2, 1)$')
plt.xlabel('p')
plt.ylabel('Density')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("beta15_1.pdf")  # Export for LaTeX
plt.show()
```

```python
fig,axes = plt.subplots(2, 2, figsize=(8,6))
x = np.linspace(0, 1, 1000)
head = [5, 54, 498, 5006]
for i in range(4):
    N = 10**(i+1)
    heads = head[i]
    newPost = beta.pdf(x, heads+2, N+1-heads)

    ax = axes[i // 2, i%2]
    ax.plot(x, newPost)
    ax.set_title(f'N = {N}, Heads = {heads}')
    ax.set_xlabel('p')
    ax.set_ylabel('Density')

plt.tight_layout()
plt.savefig("posts.pdf")
plt.show()
```
