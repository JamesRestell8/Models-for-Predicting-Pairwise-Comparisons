```python
import numpy as np
import pandas as pd
import matplotlib as plt
import math
import random
import os
import warnings

warnings.simplefilter(action="ignore", category=FutureWarning)
warnings.simplefilter(action="ignore", category=UserWarning)
warnings.simplefilter("ignore", pd.errors.SettingWithCopyWarning)
```

```python
folder = os.fsencode("../Code/seasons/BettingSeasons")
dataframes = []
for file in sorted(os.listdir(folder)):
    filename = os.fsdecode(file)
    if filename.endswith(".csv"):
        print(filename)
        df = pd.read_csv("../Code/seasons/BettingSeasons/" + filename)
        dataframes.append(df)

data = pd.concat(dataframes, axis=0)
```

```python
data = data[[
    "Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG", "FTR", "B365H", "B365D", "B365A"
]]
```

```python
data["HomeWin"] = np.where(data["FTR"] == "H", 1, 0)
data["Draw"] = np.where(data["FTR"] == "D", 1, 0)
data["AwayWin"] = np.where(data["FTR"] == "A", 1, 0)
```
